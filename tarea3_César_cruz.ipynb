##### ¿Cómo varía el monto promedio de pago por tipo de método (payment_type) según la categoría del tamaño de la transacción (creada a partir del porcentaje que representa cada compra frente al pago máximo de la plataforma)?

* Se lo pedí a gemini, lo hice como entendí jeje

```python
df_20= payments
max_pago = df_20["payment_value"].max()
def clasificar_porcentaje(porcentaje):
    if porcentaje <= 5:
        return "compra pequeña"
    elif porcentaje <= 15:
        return "compra mediana"
    else:
        return "compra grande"
df_20["categoria_transaccion"] = df_20["payment_value"].apply(
    lambda x: clasificar_porcentaje((x * 100) / max_pago))
df_20.pivot_table(
    values="payment_value",
    index="payment_type",
    columns="categoria_transaccion",
    aggfunc="mean"
)
```

```python
df_21 = order_items.merge(products, on = "product_id", how = "left")
df_21["porcentaje_flete"] = df_21.apply(
    lambda x: (x["freight_value"] / x["price"]) * 100, axis = 1)

df_21.pivot_table(
    values = ["price", "freight_value", "porcentaje_flete"],
    index = "product_category_name",
    aggfunc= "mean"
).sort_values(by = "porcentaje_flete", ascending= False)
```

```python
jeje= orders.merge(order_items, on="order_id", how="left")
df_22= jeje.merge(products, on="product_id", how="left")
def Status_resumen(x):
    if x in ["delivered","shipped"]:
        return "Entregado"
    elif x in ["canceled","unavailable"]:
        return "Cancelado/No_entregado"
    else:
        return "Procesando"
df_22["Order_status_resumido"]= df_22["order_status"].apply(Status_resumen)
df_22.pivot_table(
    values= "order_id",
    index="product_category_name",
    columns="Order_status_resumido",
    aggfunc="count"
).sort_values(by=["Procesando"],ascending=False)
```

```python
df_22["total"] = df_22.apply(
    lambda fila: fila["price"] + fila["freight_value"],
    axis=1
)
df_22.pivot_table(
    values= "total",
    index="product_category_name",
    columns="Order_status_resumido",
    aggfunc="mean"
).sort_values(by="Procesando", ascending=False)
```
