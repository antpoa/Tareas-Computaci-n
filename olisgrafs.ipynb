```python
import pandas as pd
import matplotlib.pyplot as plt
orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
sellers= pd.read_csv("olist_sellers_dataset.csv")
geo=pd.read_csv("olist_geolocation_dataset.csv")
payments=pd.read_csv("olist_order_payments_dataset.csv")

orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"],
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)
```

### Categorías de productos


```python
df_1= products. groupby("product_category_name")["product_id"].count().sort_values(ascending=False)
df_1.count()
```

```python
fig, pepe= plt.subplots(figsize=(15,5))
df_1.plot(kind="bar",
                     color=["#d18888","#5897d2","#772adc"],
                     title="Categorías",
                     xlabel="Producto",
                     ylabel="Conteo",
                     ax=pepe
                     )
plt.plot()
```

### Distribución de productos

```python
jeje=["cama_mesa_banho","esporte_lazer","moveis_decoracao","beleza_saude","utilidades_domesticas"]
def top(x):
    if x in jeje:
        return x
    else:
        return "otro"
df_3=products
df_3["product_category_name"]=df_3["product_category_name"].apply(top)
df_3= products. groupby("product_category_name")["product_id"].count().sort_values(ascending=False)
df_3
```

```python
df_3.plot(kind="pie",
          colors=["#287b9f","#c5c56a","#ae6d6d"],
          autopct="%1.3f%%",
          title="Porcentaje de productos"
          )
plt.show()
```

### Clientes por estado

```python
df_4=customers
df_4=df_4.groupby("customer_state")["customer_id"].count().sort_values(ascending=True)
df_4.plot(kind="barh",
          xlabel="Estado",
          ylabel="Nùmero de clientes",
          color="blue")
```

### Estados principales

```python
hola=["SP","RJ","MG","RS","PR"]
def topst(x):
    if x in hola:
        return x
    else:
        return "otros"
df_5=customers
df_5["customer_state"]=df_5["customer_state"].apply(topst)
df_5= df_5.groupby("customer_state")["customer_id"].count().sort_values(ascending=True)
df_5
```

```python
df_5.plot(kind="pie",
          colors=["#287b9f","#c5c56a","#ae6d6d"],
          autopct="%1.3f%%",
          title="Porcentaje de clientes por estado"
          )
plt.show()
```

### Métodos de pago

```python
df_6=payments
df_6= df_6.groupby("payment_type")["order_id"].count().sort_values(ascending=False)
df_6.plot(kind="bar",
          xlabel="metodo de pago",
          ylabel="cantidad de pagos",
          color="blue")
```

```python
nene=["credit_card","boleto","voucher","debit_card","not_defined"]
def kk(x):
    if x in nene:
        return x
    else:
        return "otros"
df_6=payments
df_6["payment_type"]=df_6["payment_type"].apply(kk)
df_6= df_6.groupby("payment_type")["order_id"].count().sort_values(ascending=True)
df_6.plot(kind="pie",
          colors=["#32383a","#c5c56a","#ae6d6d","#807373","#554747"],
          autopct="%1.3f%%",
          title="Cantidad de metodos de pago"
          )
plt.show()
```

### Comparando visualizaciones

```python
#¿Qué pregunta quiero responder? "¿Cuál utilizarías en un reporte para comparar los métodos de pago?".
#¿Qué variable o variables necesito? Necesito de la tabla de payments las columnas "payment_type" y "order_id".
#¿Qué operación debo realizar sobre los datos? Una cuenta, con el cual puedo comparar los resultados entre cada variable de "payment_type".
#¿Qué gráfica representa mejor el resultado? Una gráfica de barras, ya que con ella puedo ver la cantidad que hay entre cada uno de los pedidos dentro de los pagos.
df_8= payments
df_8= df_8.groupby("payment_type")["order_id"].count()
df_8.plot(kind="bar",
          xlabel="Cantidad de pagos",
          ylabel="Tipo de pago",
          color="green")
```
