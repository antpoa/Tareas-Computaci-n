# Manejo de columnas
Es posible seleccionar una o varias columnas:
* Una columna
* Dos columnas
* Una lista de columnas 

La sintaxis general es `df[Las columnas que quiero]`

## Ejemplo para una columna 

Para seleccionar la columna `col5` usamos `df["col5"]`

```python
# "col5" es el nombre de la columna, no sintaxis interna de python
df[["col5"]]
```

### Ejemplos para dos columnas 
Para seleccionar dos columnas la sintaxis general es `df[["col8", "col7"]]`, la diferencia con seleccionar una columna es que en esta ocasión hay doble corchete:

```python
#"col8" y "col7" son nombres de las columnas no sintaxis de python
df[["col8","col7"]]
```

### Ejemplos para seleccionar varias columnas
En general para seleccionar varias columnas es necesario escribir la *lista* de las columnas y pasarlas en corchetes al dataframe, por ejemplo, si queremos las columnas `col1`, `col10`, `col6` y `col8`, podemos utilizar la siguiente sintaxis.

```python
columnas_seleccionadas=["col1","col10", "col6", "col8"]
df[columnas_seleccionadas]
```

#   Filtrar Datos
AL crear una condición sobre una columna estamos creando ua máscara lógica (una columna de True o False por cada fila que cumpla la condición ), por ejemplo `df["col6"]=="Sur"` crea una columna que será True en aquellas filas donde el valor sea "Sur" y False cuando no lo sea.
Tenemos diferentes tipos de condiciones: 
* Condiciones numéricas.
* Condiciones categóricas.
* Varias condiciones.

Para acceder a la tabla completa en la que se cumpla la condición `df["col6"] == "Sur"` sólo basta agregarlo entre corchetes al detaframe: `df[df["col6"] == "Sur"]`. EL resultado contiene las filas y columnas `col6` tenga el valor "Sur"

```python
print("Condición categórica")
display(df[df["col6"] == "Sur"].head())
```

```python
print("Condición numérica")
display(df[df["col8"]==5].head())
```

### Filtrar por más de una condición 
Para  filtrar por más de una condición basta con separar las condiciones por paréntesis  y conectarlos con los operadores lógicos `&` (and), `|` (or) `~` (not), ejemplo: 
* Las filas cuya columna 6 sea "Sur" y columna 5 diferente de "B"

    * `(df["col6"]=="Sur") & (df["col5"] != "B")`
* Las filas cuya columna 5 sea "A" o columna 6 sea "Norte"

    * `(df["col5"]=="A") | (df["col6"]=="Norte")`

```python
df[(df["col6"]=="Sur")&(df["col5"] != "B")]
```

# Ordenar datos.
Para ordenar datos usamos `.sort_values()` y colocamos dentro de los paréntesis la columna que deseamos ordenar de menor a mayor.

```python
df.sort_values("col4")
```

También podemos odernar más de una columna, es tan simple como solo separarlas por comas y encerrarlas en corchetes `.sort_values(["col1","col2"])` y despúes si gustas ordenarlo de mayor a menor, puedes usar `.sort_values("col1", ascending= False)`.

```python
df.sort_values(["col3","col7"])
```

```python
df.sort_values("col7",ascending= False)
```

# Estadística descriptivas
Para obtener la estadística de una columna basta con seleccionar la columna y aplicar el método, por ejemplo la media es `.mean()`, el máximo es `.max()`, la mediana `.median()` y la desviación estandar `.std()`.
* `df["col1"].mean()` calcula la media (promedio aritmético) de la columna 1
* `df["col1"].max`

```python
print("Promedio:")
display(df["col1"].mean())
print("Máximo:")
display(df["col1"].max())
print("Mediana")
display(df["col1"].median())
print("Desviación estándar")
display(df["col1"].std())
```

```python
print("Descripciones")
display(df["col1"].describe())
```

# Variables categóricas.
Tenemos diferentes categorías en nuestras variables, son las siguientes:

# Group by
Con el groupby podemos hacer muchas cosas, con esta podemos realizar diferentes soluciones, `.groupby()` la primera columna que selecciones con el groupby será la columna con la que agrupes las categorías, las siguientes que selecciones puedes usarlas con las funciones que se dieron en la `Estadística descriptiva`.
Tu columna que se vaya a agrupar irá dentro de tus paréntesis y las siguientes columnas se irán poniendo dentro de unos corchetes `.groupby("col1")["col2"]` y cuando agregues más de 2 columnas entonces vuelves a encerrar en otro corchetes `.groupby("col1")[["col2","col4"]].mean()`.

```python
df.groupby("col6")["col8"].mean()
```

```python
df.groupby("col5")[["col4","col6"]].min()
```
